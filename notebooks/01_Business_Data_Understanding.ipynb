# Problem Statement

## Objective

The goal of this project is to **predict individual medical insurance charges** based on a person's demographic and health-related information.

The prediction model estimates the expected insurance cost using the following features:

- **Age** – Age of the insured individual.
- **Sex** – Gender of the individual (Male or Female).
- **BMI (Body Mass Index)** – A measure of body fat based on height and weight.
- **Children** – Number of dependents covered by the insurance plan.
- **Smoker** – Whether the individual is a smoker or a non-smoker.
- **Region** – Residential area in the United States (Northeast, Northwest, Southeast, or Southwest).

### Target Variable

- **Charges** – The individual's annual medical insurance cost.



## Machine Learning Task

This is a **Supervised Machine Learning** problem because the dataset contains labeled examples.

Since the target variable (**Charges**) is a continuous numerical value, the task is a **Regression Problem**.



## Input Features

| Feature | Description | Data Type |
|---------|-------------|-----------|
| Age | Age of the insured individual | Numerical |
| Sex | Gender (Male/Female) | Categorical |
| BMI | Body Mass Index | Numerical |

| Smoker | Smoking status (Yes/No) | Categorical |

## Target

# The impact of the region on cost

## 1. Cost of Living & Medical Service Price Variations

The cost of living varies from one city to another, which directly impacts the medical sector:
* **Operational Expenses:** Hospitals in major cities or affluent areas pay higher rent, offer higher compensation to doctors and nurses, and operate medical equipment with elevated running costs.
* **Impact:** A doctor's consultation or surgical procedure for the exact same condition will be more expensive in certain regions (e.g., the *US Northeast*) compared to others (e.g., the *Southwest*).


## 2. Demographics, Lifestyle & Health Trends

Regional culture and climate significantly influence public health:
* **Dietary Habits:** Certain areas have a higher prevalence of fast food and high-fat diets, leading to increased rates of obesity and cardiovascular diseases.
* **Environmental Factors:** Other regions may suffer from industrial air pollution, resulting in higher occurrences of asthma and respiratory allergies.

> **Insurance Impact:** Insurance providers analyze these demographics and health trends; if "Region A" records higher disease incidence rates, insurance coverage costs for its residents automatically increase.

# Smoking and Medical Insurance Costs

## 1. Why Does Smoking Increase Medical Expenses? (Cause and Effect)

Smoking is one of the strongest risk factors for many chronic and expensive diseases, including:

- Heart disease
- Cardiovascular disease
- Stroke
- Lung cancer

As a result, smokers are more likely to:

- Visit hospitals more frequently.
- Require surgeries and specialized medical treatments.
- Need long-term medications and continuous healthcare.

For insurance companies, these health conditions lead to **higher medical claims**, which significantly increase healthcare costs.



## 2. Insurance Premium Pricing (Premium Surcharge)

Insurance companies calculate premiums based on the expected level of health risk.

### Non-Smoker
- Considered a **low-risk** customer.
- Pays a **lower insurance premium** because lower medical expenses are expected.

### Smoker
- Considered a **high-risk** customer.
- Pays a **higher insurance premium** because future medical costs are expected to be significantly greater.

In many countries, such as the United States, smokers may pay **up to 50% higher insurance premiums** than non-smokers.



## 3. What Does the Medical Cost Dataset Show?

The **Medical Cost Dataset** reflects this real-world relationship between smoking and healthcare expenses.

### Key Findings

- Smokers have significantly higher medical charges than non-smokers.
- Smoking is one of the most influential features affecting medical insurance costs.
- Even after accounting for factors such as **age**, **BMI**, and **number of children**, smokers still tend to incur much higher medical expenses.
- These findings explain why insurance companies classify smokers as **high-risk customers** and charge them higher premiums.

<p style="color:red;font-size:50px">[3] Age

#  Age: "Higher Age, Higher Risk"

* **Biological Foundation:** Biologically, as age increases, immunity naturally weakens, and the likelihood of chronic diseases (such as hypertension and diabetes) or joint degeneration rises.
* **Financial Impact:** A 60-year-old individual pays a significantly higher insurance premium than a 20-year-old, as statistical projections confirm that older adults visit hospitals at a much higher rate.

<p style="color:red;font-size:50px">[4] sex

# Gender

* **Biological Foundation:** Medical costs vary between men and women based on biological differences and life stages.
* **Financial Impact:** In younger age groups (ages 18 to 45), women generally incur higher medical expenses due to pregnancy, childbirth, and specialized gynecological care. In older age groups, costs tend to equalize, or men may incur higher expenses due to a greater incidence of acute cardiovascular conditions.

<p style="color:red;font-size:50px">[5] BMI

# Body Mass Index (BMI): Obesity & Associated Complications

* **Biological Foundation:** BMI is calculated by dividing weight by height squared ($BMI = \frac{\text{weight}}{\text{height}^2}$). A value over 30 indicates obesity.
* **Financial Impact:** Obesity is not merely cosmetic; it is a primary driver of heart disease, type 2 diabetes, and back/knee issues. Consequently, an individual with a high BMI is classified as a "high-risk" policyholder, leading to higher insurance premiums—especially if combined with smoking.

<p style="color:red;font-size:50px">[6] Children

# Children / Family Size Covered

* **Biological Foundation:** The number of children determines the total number of beneficiaries covered under the same insurance policy.
* **Financial Impact:** As the number of children increases, the overall expected medical expenses rise (due to routine vaccinations, common illnesses, pediatric check-ups, and sports injuries). Consequently, the overall policy premium increases based on the number of covered individuals within the family, rather than individual health risk.
